## V42 K=197 — Multi-Paradigm Consensus Union

**Public LB:** 71.69811  
**Approach:** 4-paradigm consensus union (V35 9-model rank-avg + V39 CoilID cyclic + V40 defect-proximity + V41 BBSE) + V4_154 anchor → vote count → K-sweep.

See `APPROACH.md` for full methodology.

## Step 1 — Train the 4 paradigms

Each paradigm generates `oof_v{XX}.parquet` and `test_proba_v{XX}.parquet`.  
All use 5-fold StratifiedKFold seed=42, fold-isolated medians for leak-safety.

```bash
python train_v35.py   # 9-model rank-average, 105-feature base
python train_v39.py   # CoilID + sin/cos cyclic features
python train_v40.py   # Defect-proximity (CV-safe)
python train_v41.py   # BBSE-reweighted (w₁=9.0, w₀=0.58)
```

````python
# Train all 4 paradigms (run sequentially or in parallel)
import subprocess
for script in ['train_v35.py', 'train_v39.py', 'train_v40.py', 'train_v41.py']:
    print(f'Running {script}...')
    subprocess.run(['python', script], check=True)
````

## Step 2 — Consensus Union + K-Sweep

Run `consensus_union_v42.py` which:
1. Loads `test_proba_v{35,39,40,41}.parquet`
2. Computes top-181 per paradigm + V4_154 anchor
3. Votes per CoilID (max=5)
4. Sorts by (vote desc, mean rank-pct desc)
5. Generates K-sweep submissions

````python
import subprocess
subprocess.run(['python', 'consensus_union_v42.py'], check=True)
````
